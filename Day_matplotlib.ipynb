```python
import matplotlib.pyplot as plt
months=['jan','Feb', 'Mar', 'Apr', 'May']
sales=[1000,150,120,180,200]
plt.plot(months, sales)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ['Vikky', 'Rahul', 'Aman', 'Neha']
marks = [85,90,70,95]

plt.bar(students, marks, width=0.6)
plt.title("Students Marks")
plt.xlabel("students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#Horizontal
plt.barh(students, marks)
plt.show()
```

```python
#Multiple Bar Chart
students = ["A","B","C","D"]

python = [80,70,90,60]
java = [70,75,85,80]

import numpy as np

x = np.arange(len(students))
width = 0.35

plt.bar(x - width/2 , python,width,label="Python")
plt.bar(x + width/2 , java,width,label="Java")

plt.xticks(x,students)

plt.legend()
plt.show()
```

```python
# Pie Chart
import matplotlib.pyplot as plt

subjects = ["AI", "Python", "Java", "SQL"]
hours = [85, 90, 70, 95]
#explode = (0.1, 0, 0, 0)  # explode 1st slice
#mycolors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']

plt.pie(hours, labels=subjects, autopct='%1.1f%%', startangle=90)


plt.title("Students Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["AI","Python","Java","SQL"]
hours = [35,25,20,20]
explode = [0.1,0,0,0]
plt.pie(hours,
        labels=students,
        autopct='%1.1f%%',
        explode=explode,
        )
plt.show()
```

```python
# Scatter Plot
import matplotlib.pyplot as plt

height = [150, 160, 170, 180, 190]
weight = [50, 60, 70, 80, 90]

plt.scatter(height, weight,
             color='blue', marker='o')

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Height vs Weight")
plt.show()
```

```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt;
```

```python
data = pd.read_csv("Titanic-Dataset.csv")
```

```python
avg_fare=data.groupby("Pclass")["Fare"].mean()
plt.plot(avg_fare.index,avg_fare.values,marker="o")
plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")
plt.show()
```

```python
passengers = data["Pclass"].value_counts().sort_index()

print(passengers.index, passengers.values)
```

```python
passenger = data["Pclass"].value_counts().sort_index()

plt.bar(passengers.index, passengers.values)

plt.title("Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passenger")

plt.show()
```

```python
plt.hist(data["Age"].dropna(), bins=20)
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```
