```python
import matplotlib.pyplot as plt
months=['jan','Feb', 'Mar', 'Apr', 'May']
sales=[1000,150,120,180,200]
plt.plot(months, sales)
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

```python
#Change Line Style
plt.plot(
    months, 
    sales,
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o"
)
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales_2025 = [100, 150, 120, 180, 220]
sales_2026 = [120, 170, 160, 200, 250]

plt.plot(months, sales_2025, marker="o")
plt.plot(months, sales_2026, marker="s")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
```

```python
import matplotlib.pyplot as plt

students = ['Vikky', 'Rahul', 'Aman', 'Neha']
marks = [85,90,70,95]

plt.bar(students, marks, width=0.6)
plt.title("Students Marks")
plt.xlabel("students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#Horizontal
plt.barh(students, marks)
plt.show()
```
