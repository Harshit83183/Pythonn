```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
print(sns.__version__)
```

```python
sns.get_dataset_names()
iris = sns.load_dataset("iris")
titanic = sns.load_dataset("titanic")
penguins = sns.load_dataset("penguins")
flights = sns.load_dataset("flights")
tips = sns.load_dataset("tips")
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.show()
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
plt.show()
```

```python
flights.head()
```

```python
sns.lineplot(
    data=titanic,
    x="age",
    y="fare",
    hue="survived"
)

plt.show()
```

```python
sns.scatterplot(
    data=flights,
    x="year",
    y="passengers"
)

plt.show()
```

```python
sns.countplot(
    data=iris,
    x="species"
)
plt.show() 
```

```python
sns.countplot(
    data=iris,
    x="species"
)
plt.show()
```

```python
sns.countplot(
    data=tips,
    x="sex",
    hue="time"
    )
```

```python
sns.histplot(
    data=tips,
    x="total_bill"
)
plt.show()
```

```python
sns.histplot(
    data=tips,
    x="total_bill",
    kde=True
)
plt.show()
```
